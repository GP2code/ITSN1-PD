## Imports

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import json
```

## Age Distribution Plots

```python
def age_plots(covariate_path, ancestry, all_ancestry=False, exclude_group=["CAH"], save_plot=False):
    FIG_DIR="/home/jupyter/121825_ITSN1_Results/Figures"
    
    # load in ancestry color palette 
    
    with open(f'{FIG_DIR}/husl_ancestry_palette.json', 'r') as file:
        palette = json.load(file)
    bins = np.arange(0, 110, 5)

    df = pd.read_csv(covariate_path, sep="\t")
    sns.set(style="whitegrid")
    if all_ancestry==False:
        g = sns.displot(
        data=df,
        x="AGE",
        col="PHENO",
        hue="PHENO",
        kind="hist",
        bins=bins,
        palette=palette,
        element="step",
        stat="count",      
        common_norm=False,
        facet_kws={"sharey": False},
        col_wrap=2            # if you want them side-by-side
    )
    elif all_ancestry==True:
        # remove groups we want to exclude from figure
        
        df = df[~(df["Ancestry"].isin(exclude_group))]
        
        
        g = sns.displot(
        data=df,
        x="AGE",
        col="PHENO",
        hue="Ancestry", # Make hue ancestry instead so we can have all ancestries on one graph
        kind="hist",
        bins=bins,
        palette=palette,
        hue_order=sorted(df["Ancestry"].unique()),
        fill=False,
        element="step",
        stat="count",      
        common_norm=False,
        facet_kws={"sharey": False},
        col_wrap=2            # if you want them side-by-side
    )
    g.fig.suptitle(
    f"Age distribution in All of Us Cohort by phenotype: {ancestry}",
    fontsize=14,
    y=1.05)
    g.set_axis_labels("Age (years)", "Count")
    g.set_titles("PHENO: {col_name}")
    title_map = {
    1: "Controls",
    2: "PD Cases"
    }
    
   
    for ax, pheno in zip(g.axes.flat, g.col_names):
        ax.set_title(f"{title_map.get(pheno, pheno)}")
    
    
    # save figures to file
    
    if save_plot & all_ancestry:
        g.savefig(f"{FIG_DIR}/aou_age_distribution_plot_for_itsn1_proj.png")
    elif save_plot:
        g.savefig(f"{FIG_DIR}/{ancestry}_age_distribution_plot.png")
    plt.show()
```

```python
import warnings

warnings.filterwarnings(
    "ignore",
    category=FutureWarning,
    module="seaborn"
)
```

```python
# run function to generate plots per ancestry
for ancestry in ['AAC','AJ','AFR','AMR','CAS','EAS','EUR','FIN','MDE','SAS']:
    WORK_DIR = f"/home/jupyter/121825_ITSN1_Results/{ancestry}_R8_wgs"
    covariate_path = f"{WORK_DIR}/{ancestry}_covariate_file.txt"
    age_plots(covariate_path, ancestry)  
```

```python
# run function to generate plots with all ancestries on one panel
WORK_DIR = f"/home/jupyter/121825_ITSN1_Results/"
covariate_path = f"{WORK_DIR}/data/covariate_wgs_R8.txt"


age_plots(covariate_path, "All Ancestry", all_ancestry=True, exclude_group=["CAH", "CAS", "EAS", "FIN", "MDE", "SAS"], save_plot=True)
 
```
